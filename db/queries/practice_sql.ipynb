### Thư viện 

```python
import pandas as pd
from sqlalchemy import create_engine
import os
from dotenv import load_dotenv
```

### Load đường dẫn

```python
load_dotenv()
DATABASE_URL = os.getenv("DATABASE_URL")
engine = create_engine(DATABASE_URL)
```

### SQL practice

```python
quey = """
select * from sales
where region = 'South' and sale_date < '2024-01-11'
"""

df = pd.read_sql(quey, engine)
print(df)
```

```python
query = """
select customer_name, SUM(amount) as total_amount from sales
group by customer_name
having SUM(amount) > 1000
"""

df = pd.read_sql(query, engine)
print(df)
```

```python
query = """
SELECT COUNT(*) as total_sales, customer_name FROM sales
group by customer_name
""" 

df = pd.read_sql(query, engine)
print(df)
```

```python
query = """
select SUM(amount) as total_amount, customer_name from sales
group by customer_name
"""

df = pd.read_sql(query, engine)
print(df)
```

```python
query = """
select region, sum(amount) as total_sales, avg(amount) as average_sales, sum(amount) - avg(amount) as difference
from sales
group by region
"""

df = pd.read_sql(query, engine)
print(df)


```

```python
query = """
with total_amounts as (
    select customer_name, sum(amount) as total_amount
    from sales
    group by customer_name
) 

select s.customer_name, total_amount, amount, amount/total_amount as percentage
from sales s
left join total_amounts 
on s.customer_name = total_amounts.customer_name

"""

df = pd.read_sql(query, engine)
print(df)

```

```python
query = """
SELECT s.customer_name
FROM sales s
left join customers c
on s.customer_name = c.customer_name
WHERE c.customer_name IS NULL
"""

df = pd.read_sql(query, engine)
print(df)

```
